```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "Name":   ["Ali", "Sara", "John", "Mia", "Omar", "Zara", "Tom", "Nina", "Lena"],
    "Age":    [20, 22, 21, 21, 23, 20, 22, 21, 21],
    "Gender": ["M", "F", "M", "F", "M", "F", "M", "F", "F"],
    "Score":  [85, 90, 78, 95, 83, 88, 72, 91, 60],
    "Grade":  ["B", "A", "C", "A", "B", "B", "C", "A", "D"]
}
df = pd.DataFrame(data)
print("✅ Data ready!")
print(df)
```

```python
plt.figure(figsize=(10,5))
plt.bar(df["Name"], df["Score"], color="steelblue")
plt.title("Student Scores")
plt.xlabel("Student")
plt.ylabel("Score")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
grade_counts = df["Grade"].value_counts()
plt.figure(figsize=(6,6))
plt.pie(grade_counts, labels=grade_counts.index, autopct="%1.1f%%", colors=["green","blue","orange","red"])
plt.title("Grade Distribution")
plt.show()
```

```python
gender_avg = df.groupby("Gender")["Score"].mean()
plt.figure(figsize=(5,4))
plt.bar(gender_avg.index, gender_avg.values, color=["skyblue","pink"])
plt.title("Average Score by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Score")
plt.show()
```

```python
print("📊 KEY INSIGHTS:")
print("Highest Score:", df["Score"].max(), "— by", df.loc[df["Score"].idxmax(), "Name"])
print("Lowest Score:", df["Score"].min(), "— by", df.loc[df["Score"].idxmin(), "Name"])
print("Average Score:", round(df["Score"].mean(), 2))
print("Top Grade (A) students:", list(df[df["Grade"]=="A"]["Name"]))
```

```python
from sklearn.linear_model import LinearRegression
import numpy as np

print("✅ Library ready!")
```

```python
# X = Age (input), Y = Score (output)
X = df[["Age"]]
Y = df["Score"]

model = LinearRegression()
model.fit(X, Y)

print("✅ Model trained!")
print("Score prediction formula ready!")
```

```python
# Agar kisi student ki age 21 hai to score kya hoga?
age_to_predict = [[21]]
predicted_score = model.predict(age_to_predict)

print(f"Age 21 ka predicted score: {round(predicted_score[0], 2)}")
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Age"], df["Score"], color="steelblue", label="Actual Scores")
plt.plot(df["Age"], model.predict(X), color="red", label="Prediction Line")
plt.title("Age vs Score — Prediction Model")
plt.xlabel("Age")
plt.ylabel("Score")
plt.legend()
plt.show()
```

```python
print("📊 MODEL INSIGHTS:")
print("R² Score:", round(model.score(X, Y), 2))
print("(R² = 1.0 means perfect prediction, 0 means no relation)")
```
