```python
# Install pandas (tool for working with data)
import pandas as pd

# Create a sample student dataset
data = {
    "Name":    ["Ali", "Sara", "John", "Mia", "Omar", "Zara", "Tom", "Nina", "Ali", "Lena"],
    "Age":     [20, 22, 21, None, 23, 20, 22, None, 20, 21],
    "Gender":  ["M", "F", "M", "F", "M", "F", "M", "F", "M", "F"],
    "Score":   [85, 90, 78, 95, None, 88, 72, 91, 85, 60],
    "Grade":   ["B", "A", "C", "A", "B", "B", "C", "A", "B", "D"]
}

df = pd.DataFrame(data)
print("✅ Dataset loaded!")
print(df)
```

```python
# How many rows and columns?
print("Shape:", df.shape)
```

```python
# What are the column names and data types?
print(df.dtypes)
```

```python
# See a summary of the data
df.info()
```

```python
# Basic statistics (min, max, average...)
df.describe()
```

```python
# Step 1: Check for missing values
print("Missing values:\n", df.isnull().sum())
```

```python
# Step 2: Fill missing Age with the average age
df["Age"] = df["Age"].fillna(df["Age"].mean())
print("Age missing values fixed!")
```

```python
# Step 3: Fill missing Score with the average score
df["Score"] = df["Score"].fillna(df["Score"].mean())
print("Score missing values fixed!")
```

```python
# Step 4: Remove duplicate rows
df = df.drop_duplicates()
print("Duplicates removed! Rows now:", len(df))
```

```python
# Step 5: Confirm everything is clean
print("Missing values after cleaning:\n", df.isnull().sum())
print(df)
```

```python
# Step 1: Basic statistics
print("Average Score:", df["Score"].mean())
print("Highest Score:", df["Score"].max())
print("Lowest Score:", df["Score"].min())
```

```python
# Step 2: Count grades
print("\nGrade Distribution:")
print(df["Grade"].value_counts())
```

```python
# Step 3: Average score by gender
print("\nAverage Score by Gender:")
print(df.groupby("Gender")["Score"].mean())
```

```python
# Step 4: Visualize scores with a chart
import matplotlib.pyplot as plt

plt.bar(df["Name"], df["Score"], color="steelblue")
plt.title("Student Scores")
plt.xlabel("Student")
plt.ylabel("Score")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Step 5: Spot outliers — who scored very low?
low_scorers = df[df["Score"] < 70]
print("Students scoring below 70:\n", low_scorers)
```
